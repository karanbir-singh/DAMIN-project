```python
import pandas as pd
```

```python
df = pd.read_csv("../data/spotify_no_null.csv")

genres = df["track_genre"].unique()
genres
```

```python
def group_genres(genres):
    categories = {
        "rock": ["rock", "metal"],
        
        "pop": ["pop"],
            
        "latin": ["latin", "reggaeton"],
            
        "classical": ["classical", "opera", "piano"],
            
        "dance": ["dance", "electro", "house", "techno", "disco"]
    }

    # Inizializza i gruppi
    grouped = {key: [] for key in categories}
    grouped["other"] = []

    # Matching dei generi
    for g in genres:
        found = False
        for cat, items in categories.items():
            if any(k in g for k in items):
                grouped[cat].append(g)
                found = True
                break
        if not found:
            grouped["other"].append(g)

    return grouped
```

```python
grouped = group_genres(genres)

for cat, items in grouped.items():
    print(cat, "=>", items)
```

```python
mapped_genres = []
for genre in df["track_genre"]:
    for cat, items in grouped.items():
        if genre in items:
            mapped_genres.append(cat)

df["track_genre"] = mapped_genres
```
